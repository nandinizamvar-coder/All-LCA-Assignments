```python

import pandas as pd
data = {
    "Student": ["Amit", "Bhavna", "Chetan", "Diya", "Esha", "Farhan", "Amit", "Gauri"],
    "Maths": [75, 82, None, 90, 72, 85, 75, 68],
    "Python": [80, 78, 70, 92, None, 88, 80, 74],
    "Attendance": [85, 90, 78, 95, 80, 92, 85, None],
    "City": ["Pune", "Mumbai", "Pune", "Nashik", "Mumbai", "Pune", "Pune", "Mumbai"],
    "Remarks": ["Good", "Good", "Average", "Good", "Average", "Good", "Good", "Average"]
}
df = pd.DataFrame(data)
df.to_csv("assignment_06.csv", index=False)
print("CSV file created successfully!")
```

```python

import pandas as pd
df = pd.read_csv("assignment_06.csv")

print("Original Dataset:")
print(df)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDataset after removing missing values:")
print(df.dropna())

df["Maths"] = df["Maths"].fillna(df["Maths"].mean())
df["Python"] = df["Python"].fillna(df["Python"].mean())
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].mean())
df["City"] = df["City"].fillna(df["City"].mode()[0])

print("\nDataset after filling missing values:")
print(df)

print("\nDuplicate Rows:")
print(df[df.duplicated()])

df = df.drop_duplicates()

print("\nDataset after removing duplicates:")
print(df)

df = df.drop("Remarks", axis=1)

print("\nDataset after removing Remarks column:")
print(df)

print("\nStudents with Maths marks 70 or above:")
print(df.loc[df["Maths"] >= 70])

df.to_csv("cleaned_data_assignment_06.csv", index=False)

print("\nCleaned data saved successfully.")
```
