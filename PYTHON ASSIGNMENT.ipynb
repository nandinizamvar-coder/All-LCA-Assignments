```python
# Assignment 1: Student Data Management Using Python Collections
```

```python
# Creating student records
students = {}

students[101] = ("Ananya", [85, 90, 88])
students[102] = ("Riya", [78, 82, 80])
students[103] = ("Kabir", [92, 89, 94])

# Displaying student records
print("Student Records:")
for student_id, details in students.items():
    print("Student ID:", student_id)
    print("Name:", details[0])
    print("Marks:", details[1])
    print()

# Adding a new student
students[104] = ("Neha", [86, 91, 84])
print("After adding new student:")
print(students)

# Updating marks
students[102] = ("Riya", [82, 85, 88])
print("\nAfter updating Riya's marks:")
print(students)

# Deleting a student
del students[103]
print("\nAfter deleting student 103:")
print(students)

# Calculating average marks
print("\nAverage Marks:")
for student_id, details in students.items():
    marks = details[1]
    average = sum(marks) / len(marks)
    print(details[0], "-", average)

# Final student records
print("\nFinal Student Records:")
for student_id, details in students.items():
    print(student_id, "-", details[0], "-", details[1])
```

```python
# Find the largest of three numbers

a = 25
b = 42
c = 18

if a >= b and a >= c:
    largest = a
elif b >= a and b >= c:
    largest = b
else:
    largest = c

print("The largest number is:", largest)
```

```python
# Assignment 3: Check Right-Angled Triangle Using Function
```

```python
# Function to check whether a triangle is right-angled

def check_right_angle(a, b, c):
    sides = sorted([a, b, c])
    
    if sides[0]**2 + sides[1]**2 == sides[2]**2:
        print("The triangle is right-angled.")
    else:
        print("The triangle is not right-angled.")

check_right_angle(3, 4, 5)
```
