```python
# Grocery Price Tracker
```

```python
# Store the prices of grocery items in a list
prices = [250, 650, 450, 700, 300]

# Remove items priced above ₹500
prices = [price for price in prices if price <= 500]
print

# Insert a new item price at position 2
prices.insert(2, 400)

# Sort the list in ascending order
prices.sort()

# Calculate total cost
total_cost = sum(prices)

# Display the updated list and total cost
print("Updated grocery prices:", prices)
print("Total cost:", total_cost)
```

```python
# Student Records using Tuple
```

```python
# Store student information in a tuple
students = ("Amit", 19, "A", "Neha", "B")

# Count how many times grade "A" appears
grade_count = students.count("A")

# Find the index of the first occurrence of "A"
grade_index = students.index("A")

# Update Neha's grade from "B" to "A"
students = ("Amit", 19, "A", "Neha", "A")

# Display the results
print("Original student tuple:", ("Amit", 19, "A", "Neha", "B"))
print("Number of times grade A appears:", grade_count)
print("Index of first A grade:", grade_index)
print("Updated student tuple:", students)
```

```python
# Travel Itinerary using List and Tuple
```

```python
# Store a list of cities
cities = ["Mumbai", "Delhi", "Jaipur"]

# Add two more cities
cities.append("Agra")
cities.append("Goa")

# Remove a city by name
cities.remove("Delhi")

# Convert the list into a tuple
travel_itinerary = tuple(cities)

# Access first, last and middle city
first_city = travel_itinerary[0]
last_city = travel_itinerary[-1]
middle_city = travel_itinerary[len(travel_itinerary) // 2]

# Display the results
print("Travel itinerary:", travel_itinerary)
print("First city:", first_city)
print("Last city:", last_city)
print("Middle city:", middle_city)
```

```python
# Sports Tournament Results using List and Tuple
```

```python
# Store the scores of 5 players
scores = [78, 92, 65, 88, 75]

# Sort the scores in descending order
scores.sort(reverse=True)

# Convert the list into a tuple
score_tuple = tuple(scores)

# Unpack the tuple into separate variables
score1, score2, score3, score4, score5 = score_tuple

# Calculate the average score
average_score = sum(score_tuple) / len(score_tuple)

# Display the results
print("Scores in descending order:", score_tuple)
print("Highest score:", score1)
print("Lowest score:", score5)
print("Average score:", average_score)
```

```python
# Student Records Management
```

```python
# 1. Create a list of students
students = ["Amir", "Neha", "Rahul", "Sneha", "Rahul"]
print("Original list:", students)

# 2. Append Vikas
students.append("Vikas")
print("After appending Vikas:", students)

# 3. Insert Priya at position 2
students.insert(2, "Priya")
print("After inserting Priya:", students)

# 4. Count occurrences of Rahul
rahul_count = students.count("Rahul")
print("Rahul appears", rahul_count, "times")

# 5. Find index of Neha
neha_index = students.index("Neha")
print("Index of Neha:", neha_index)

# 6. Remove first occurrence of Rahul
students.remove("Rahul")
print("After removing first Rahul:", students)

# 7. Remove the last student
removed_student = students.pop()
print("Removed student:", removed_student)
print("After pop:", students)

# 8. Make a copy of the list
backup_students = students.copy()
print("Backup students:", backup_students)

# 9. Extend with Karan and Meena
students.extend(["Karan", "Meena"])
print("After extending:", students)

# 10. Sort alphabetically
students.sort()
print("After sorting:", students)

# 11. Reverse the list
students.reverse()
print("After reversing:", students)
```

```python
# Exercise 7 - Count Characters in a String

text = input("Enter a string: ")

count = {}

for char in text:
    if char in count:
        count[char] += 1
    else:
        count[char] = 1

print("Character count:", count)
```

```python
# Exercise 8 - Quotient and Remainder Using Tuple

numerator = int(input("Enter numerator: "))
denominator = int(input("Enter denominator: "))

quotient = numerator // denominator
remainder = numerator % denominator

result = (quotient, remainder)

print("Quotient and Remainder:", result)
```

```python
# Exercise 9 - Number of Days Between Two Dates

d1 = int(input("Enter day of first date: "))
m1 = int(input("Enter month of first date: "))
y1 = int(input("Enter year of first date: "))

d2 = int(input("Enter day of second date: "))
m2 = int(input("Enter month of second date: "))
y2 = int(input("Enter year of second date: "))

days = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

total1 = y1 * 365
for i in range(m1 - 1):
    total1 += days[i]
total1 += d1

total2 = y2 * 365
for i in range(m2 - 1):
    total2 += days[i]
total2 += d2

difference = total2 - total1

if difference < 0:
    difference = -difference

print("Number of days between the dates =", difference)
```
