```python
students = {
    101: ("Riya", "CSE", [85,90,88]),
    102: ("Kabir", "ECE", [78,82,80]),
    103: ("Meera", "AIDS", [91,89,94])
          }

print("Original student records:")
print(students)

#Adding a new student
students[104] = ("Arjun", "CSE", [76,81,79])
print("\nAfter adding a new student:")
print(students)

#Deleting a student
del students[102]
print("\nAfter deleting student 102:")
print(students)

# Updating a student's details
students[101] = ("Riya", "AIDS", [88,92,90])
print("\nAfter updating student 101:")
print(students)

#displaying final records
print("\nFinal student records:")

for roll_number in students:
    print("Roll Number:", roll_number)
    print("Name:", students[roll_number][0])
    print("Branch:", students[roll_number][1])
    print ("Marks:", students[roll_number][2])
    print()
```
