```python
import pandas as pd

data = {
    "Student": ["Amit", "Bhavna", "Chetan", "Diya", "Esha", "Farhan"],
    "Maths": [75, 82, 68, 90, 72, 85],
    "Python": [80, 78, 70, 92, 75, 88],
    "Attendance": [85, 90, 78, 95, 80, 92],
    "City": ["Pune", "Mumbai", "Pune", "Nashik", "Mumbai", "Pune"]
}
df = pd.DataFrame(data)
df.to_csv("assignment_07.csv", index=False)
print("CSV file created successfully!")

```

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

filename = input("Enter the CSV file name: ")

df = pd.read_csv(filename)

print("\nStudent Dataset:")
print(df)

print("\nAverage Maths Marks:", df["Maths"].mean())
print("Average Python Marks:", df["Python"].mean())
print("Average Attendance:", df["Attendance"].mean())

plt.figure(figsize=(7, 4))
plt.plot(df["Student"], df["Python"], marker="o", label="Python Marks")
plt.title("Python Marks of Students")
plt.xlabel("Student")
plt.ylabel("Python Marks")
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(7, 4))
plt.bar(df["Student"], df["Maths"], label="Maths Marks")
plt.title("Maths Marks of Students")
plt.xlabel("Student")
plt.ylabel("Maths Marks")
plt.legend()
plt.show()

plt.figure(figsize=(7, 4))
plt.scatter(df["Maths"], df["Python"], label="Students")
plt.title("Maths vs Python Marks")
plt.xlabel("Maths Marks")
plt.ylabel("Python Marks")
plt.legend()
plt.grid(True)
plt.show()

correlation = df[["Maths", "Python", "Attendance"]].corr()

print("\nCorrelation Matrix:")
print(correlation)

plt.figure(figsize=(6, 5))
sns.heatmap(correlation, annot=True, cmap="Blues", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.show()

plt.figure(figsize=(7, 4))
plt.hist(df["Python"], bins=5, edgecolor="black", label="Python Marks")
plt.title("Histogram of Python Marks")
plt.xlabel("Python Marks")
plt.ylabel("Number of Students")
plt.legend()
plt.show()

plt.figure(figsize=(7, 4))
plt.boxplot([df["Maths"], df["Python"]], tick_labels=["Maths", "Python"])
plt.title("Box Plot of Maths and Python Marks")
plt.ylabel("Marks")
plt.show()

city_count = df["City"].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(city_count, labels=city_count.index, autopct="%1.1f%%",
        startangle=90)
plt.title("Students from Each City")
plt.legend(title="City")
plt.show()

print("\nProgram executed successfully!")
```
